### Q2 · One categorical variable
**How does enrollment compare across majors?**

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

fig, ax = plt.subplots()

major_counts = students["major"].value_counts().sort_values(ascending=True)

major_counts.plot(
    kind="barh",
    color=NAVY,
    ax=ax
)

ax.set_title("Northgate Student Enrollment by Major")
ax.set_xlabel("Number of Students")
ax.set_ylabel("Major")

plt.show()
```

### Q4 · Two quantitative variables
**Is there a relationship between how far students commute and their final GPA?** Remember: *association is not causation.*
→ *scatterplot with a trend line*. Columns: `commute_miles` and `final_gpa` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

fig, ax = plt.subplots()

sns.regplot(
    data=students,
    x="commute_miles",
    y="final_gpa",
    scatter_kws={"alpha": 0.35},
    line_kws={"color": RUST},
    ax=ax
)

ax.set_title("Relationship Between Commute Distance and Final GPA")
ax.set_xlabel("Commute Distance (Miles)")
ax.set_ylabel("Final GPA")

plt.show()
```

**STAR story — Q4** ***Scope - ** This chart shows the relationship between students' commute distance in miles and their final GPA.

**Track - ** The question is whether students who commute farther tend to have different final GPAs.

**Articulate - ** The chart shows a negative relationship between commute distance and final GPA. As commute distance increases, final GPA tends to decrease. The relationship is not perfect because there is substantial variation among individual students. The correlation is approximately -0.30, which suggests a moderate negative association.

**Respond - ** The pattern is worth investigating, but association does not mean causation. A longer commute may be related to other factors such as work hours, housing, or study time. I would examine those variables next before drawing conclusions about why GPA differs.

### Q5 · Over time
**How does student engagement change across the semester?**
→ *line chart* of average weekly activity. Use the `weekly` table: `week` and `minutes_active` (you'll need to average by week first — see the Guided notebook).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

weekly_avg = (
    weekly
    .groupby("week", as_index=False)["minutes_active"]
    .mean()
)

fig, ax = plt.subplots()

sns.lineplot(
    data=weekly_avg,
    x="week",
    y="minutes_active",
    marker="o",
    color=NAVY,
    ax=ax
)

ax.set_title("Average Student Engagement Across the Semester")
ax.set_xlabel("Week")
ax.set_ylabel("Average Minutes Active")

plt.show()
```

**STAR story — Q5** ***Scope - ** This chart shows the average number of minutes students were active each week using the `weekly_activity` table.

**Track - ** The question is how student engagement changes across the semester.

**Articulate - ** Average activity is fairly stable at around 210 minutes per week for most of the semester. There is a very large drop in Week 9, when average activity falls to about 89 minutes, before returning to roughly 210 minutes in Week 10.

**Respond - ** The Week 9 drop is unusual and should be investigated. I would check whether there was a break, system outage, incomplete data collection, or another event that explains the sudden decrease.
